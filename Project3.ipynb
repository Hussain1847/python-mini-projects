# 🔢 Project 3: Number Guessing Game

Two ways to play:
1. **You guess** the computer's secret number. It tells you if you're too high or too low.
2. **The computer guesses** your number. It uses *binary search*, so it needs at most 7 tries for any number from 1 to 100.

**Concepts:** `random`, `while` loops, input validation with `try/except`, binary search.

```python
import random

MAX_NUMBER = 100  # change this to make the game easier or harder


def read_int(prompt: str, low: int, high: int) -> int:
    """Ask for a whole number between low and high (inclusive)."""
    while True:
        raw = input(prompt).strip()
        try:
            value = int(raw)
        except ValueError:
            print(f"'{raw}' is not a whole number. Try again.")
            continue
        if low <= value <= high:
            return value
        print(f"Please enter a number between {low} and {high}.")


def guess(x: int) -> int:
    """The player guesses a random number between 1 and x. Returns the number of attempts."""
    secret = random.randint(1, x)
    attempts = 0
    while True:
        attempt = read_int(f"Guess a number between 1 and {x}: ", 1, x)
        attempts += 1
        if attempt < secret:
            print("Too low. Try again.")
        elif attempt > secret:
            print("Too high. Try again.")
        else:
            print(f"🎉 Correct! The number was {secret}. You needed {attempts} attempt(s).")
            return attempts


def computer_guess(x: int) -> int | None:
    """The computer guesses the player's number using binary search."""
    low, high = 1, x
    attempts = 0
    print(f"Think of a number between 1 and {x}, and I'll guess it!")
    while low <= high:
        attempt = (low + high) // 2  # always guess the middle of what's left
        feedback = input(f"Is {attempt} too high (H), too low (L), or correct (C)? ").strip().lower()
        if feedback not in ("h", "l", "c"):
            print("Please answer with H, L or C.")
            continue
        attempts += 1
        if feedback == "c":
            print(f"🎉 Got it! Your number is {attempt}. I needed {attempts} attempt(s).")
            return attempts
        if feedback == "h":
            high = attempt - 1
        else:
            low = attempt + 1
    print("Hmm, those answers don't add up. Did you change your number? 🤔")
    return None


def main() -> None:
    print("🔢 Number Guessing Game")
    print("  1) You guess my number")
    print("  2) I guess your number")
    mode = read_int("Choose a mode (1 or 2): ", 1, 2)
    if mode == 1:
        guess(MAX_NUMBER)
    else:
        computer_guess(MAX_NUMBER)


if __name__ == "__main__":
    main()
```
