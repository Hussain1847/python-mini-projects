# 🔐 Project 7: Password Generator

Generates strong random passwords. Every password contains at least one uppercase letter, one lowercase letter, one digit and one symbol.

It uses Python's `secrets` module, which is designed for security. The `random` module is fine for games but too predictable for passwords.

**Concepts:** `secrets`, the `string` module, list building, input validation.

```python
import secrets
import string

SYMBOLS = "@#$%&*=!?"
CHARACTER_POOLS = [string.ascii_uppercase, string.ascii_lowercase, string.digits, SYMBOLS]
ALL_CHARACTERS = "".join(CHARACTER_POOLS)
MIN_LENGTH, MAX_LENGTH = 8, 128


def generate_password(length: int) -> str:
    """Return a random password with at least one character from every pool."""
    if length < len(CHARACTER_POOLS):
        raise ValueError(f"length must be at least {len(CHARACTER_POOLS)}")
    chars = [secrets.choice(pool) for pool in CHARACTER_POOLS]
    chars += [secrets.choice(ALL_CHARACTERS) for _ in range(length - len(chars))]
    secrets.SystemRandom().shuffle(chars)  # so the guaranteed characters aren't always first
    return "".join(chars)


def read_int(prompt: str, low: int, high: int) -> int:
    """Ask for a whole number between low and high (inclusive)."""
    while True:
        raw = input(prompt).strip()
        if raw.isdigit() and low <= int(raw) <= high:
            return int(raw)
        print(f"Please enter a whole number from {low} to {high}.")


def main() -> None:
    print("🔐 Welcome to the Password Generator\n")
    count = read_int("How many passwords? (1-20): ", 1, 20)
    length = read_int(f"Password length? ({MIN_LENGTH}-{MAX_LENGTH}): ", MIN_LENGTH, MAX_LENGTH)

    print("\nHere are your passwords:")
    for number in range(1, count + 1):
        print(f"  {number:>2}. {generate_password(length)}")


if __name__ == "__main__":
    main()
```
