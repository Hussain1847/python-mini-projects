# ✂️ Project 4: Rock, Paper, Scissors

Play as many rounds as you like against the computer. The score is kept until you quit.

**Rules:** rock beats scissors, scissors beats paper, paper beats rock.

**Concepts:** dictionaries as lookup tables, `random.choice`, game loops.

```python
import random

CHOICES = {"r": "Rock 🪨", "p": "Paper 📄", "s": "Scissors ✂️"}
BEATS = {"r": "s", "s": "p", "p": "r"}  # each key beats its value
MESSAGES = {"win": "You won! 🎉", "loss": "You lost! 😢", "tie": "It's a tie! 🤝"}


def is_win(player: str, opponent: str) -> bool:
    """Return True if `player` beats `opponent`."""
    return BEATS[player] == opponent


def get_player_choice() -> str:
    """Ask until the player types r, p, s or q."""
    while True:
        choice = input("Choose (r)ock, (p)aper, (s)cissors or (q)uit: ").strip().lower()
        if choice in CHOICES or choice == "q":
            return choice
        print("Invalid choice. Please type r, p, s or q.")


def play_round(player: str) -> str:
    """Play one round and return 'win', 'loss' or 'tie'."""
    computer = random.choice(list(CHOICES))
    print(f"You: {CHOICES[player]}   vs   Computer: {CHOICES[computer]}")
    if player == computer:
        return "tie"
    return "win" if is_win(player, computer) else "loss"


def main() -> None:
    score = {"win": 0, "loss": 0, "tie": 0}
    while True:
        player = get_player_choice()
        if player == "q":
            break
        result = play_round(player)
        score[result] += 1
        print(MESSAGES[result])
        print(f"Score: You {score['win']} | Computer {score['loss']} | Ties {score['tie']}\n")
    print("Thanks for playing! 👋")


if __name__ == "__main__":
    main()
```
