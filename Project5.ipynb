# 🪢 Project 5: Hangman

Guess the hidden word one letter at a time before the hangman drawing is finished. You have 7 lives.

The word list and the ASCII drawings are included in this notebook, so it runs anywhere. It no longer needs extra files on Google Drive.

**Concepts:** sets, list comprehensions, string methods, game state.

```python
WORDS = [
    "python", "variable", "function", "keyboard", "monitor", "algorithm", "notebook",
    "dictionary", "string", "integer", "boolean", "compiler", "debugger", "internet",
    "browser", "database", "network", "software", "hardware", "developer", "project",
    "library", "module", "package", "terminal", "github", "laptop", "pixel", "server",
    "cloud", "binary", "syntax", "loop", "random", "galaxy", "planet", "rocket",
    "jungle", "elephant", "giraffe", "penguin", "dolphin", "mountain", "island",
    "volcano", "thunder", "rainbow", "pyramid", "castle", "dragon", "wizard", "puzzle",
]

# One drawing for each number of lives left (index = lives).
HANGMAN_STAGES = [
    r"""
  +---+
  |   |
  X   |
 /|\  |
 / \  |
      |
=========""",
    r"""
  +---+
  |   |
  O   |
 /|\  |
 / \  |
      |
=========""",
    r"""
  +---+
  |   |
  O   |
 /|\  |
 /    |
      |
=========""",
    r"""
  +---+
  |   |
  O   |
 /|\  |
      |
      |
=========""",
    r"""
  +---+
  |   |
  O   |
 /|   |
      |
      |
=========""",
    r"""
  +---+
  |   |
  O   |
  |   |
      |
      |
=========""",
    r"""
  +---+
  |   |
  O   |
      |
      |
      |
=========""",
    r"""
  +---+
  |   |
      |
      |
      |
      |
=========""",
]
```

```python
import random
import string

MAX_LIVES = len(HANGMAN_STAGES) - 1  # 7


def get_valid_word(words: list[str]) -> str:
    """Pick a random word that has no spaces or hyphens."""
    word = random.choice(words)
    while "-" in word or " " in word:
        word = random.choice(words)
    return word.upper()


def hangman() -> None:
    word = get_valid_word(WORDS)
    word_letters = set(word)                 # letters still to be guessed
    alphabet = set(string.ascii_uppercase)
    used_letters: set[str] = set()           # every letter the player has tried
    lives = MAX_LIVES

    while word_letters and lives > 0:
        print(HANGMAN_STAGES[lives])
        print(f"Lives left: {lives}   Used letters: {' '.join(sorted(used_letters)) or '-'}")
        print("Word:", " ".join(letter if letter in used_letters else "_" for letter in word))

        user_letter = input("Guess a letter: ").strip().upper()
        if len(user_letter) != 1 or user_letter not in alphabet:
            print("\nPlease type a single letter from A to Z.")
        elif user_letter in used_letters:
            print(f"\nYou already tried {user_letter}. Guess another letter.")
        else:
            used_letters.add(user_letter)
            if user_letter in word_letters:
                word_letters.remove(user_letter)
                print(f"\nNice! {user_letter} is in the word.")
            else:
                lives -= 1
                print(f"\nSorry, {user_letter} is not in the word.")

    if lives == 0:
        print(HANGMAN_STAGES[0])
        print(f"💀 You ran out of lives. The word was {word}.")
    else:
        print(f"🎉 You guessed the word {word} with {lives} lives to spare!")


if __name__ == "__main__":
    hangman()
```
