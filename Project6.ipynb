# ⏱️ Project 6: Countdown Timer

Enter a duration in seconds (`90`) or as minutes:seconds (`1:30`). The timer counts down to zero.
Press the stop button (or `Ctrl+C`) to cancel it early.

**Concepts:** `time.sleep`, `divmod`, string formatting, parsing user input.

```python
import time


def parse_duration(text: str) -> int:
    """Turn '90' or '1:30' into a number of seconds. Raises ValueError if invalid."""
    text = text.strip()
    if ":" in text:
        minutes, seconds = text.split(":", 1)
        total = int(minutes) * 60 + int(seconds)
    else:
        total = int(text)
    if total <= 0:
        raise ValueError("duration must be positive")
    return total


def countdown(total_seconds: int) -> None:
    """Print a MM:SS countdown that updates in place once per second."""
    try:
        for remaining in range(total_seconds, 0, -1):
            mins, secs = divmod(remaining, 60)
            print(f"\r⏳ {mins:02d}:{secs:02d}", end="", flush=True)
            time.sleep(1)
        print("\r✅ Timer completed!   ")
    except KeyboardInterrupt:
        print("\n⏹️ Timer stopped.")


def main() -> None:
    while True:
        try:
            seconds = parse_duration(input("Enter the time (seconds or mm:ss): "))
            break
        except ValueError:
            print("Please enter a positive number like 45 or 2:30.")
    countdown(seconds)


if __name__ == "__main__":
    main()
```
