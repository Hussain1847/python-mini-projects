# ⚖️ Project 8: BMI Calculator

Calculates your Body Mass Index (BMI) from your weight and height and shows the matching WHO category.

| BMI | Category |
|-----|----------|
| below 18.5 | Underweight |
| 18.5 – 24.9 | Normal weight |
| 25 – 29.9 | Overweight |
| 30 and above | Obese |

*BMI is a rough screening tool, not a diagnosis.*

**Concepts:** arithmetic, `float()` conversion, `if/elif` chains, input validation.

```python
#@title BMI Calculator

def read_number(prompt: str, low: float, high: float) -> float:
    """Ask for a number in a sensible range."""
    while True:
        raw = input(prompt).strip().replace(",", ".")
        try:
            value = float(raw)
        except ValueError:
            print("Please enter a number, for example 65 or 172.5.")
            continue
        if low <= value <= high:
            return value
        print(f"Please enter a value between {low:g} and {high:g}.")


def calculate_bmi(weight_kg: float, height_m: float) -> float:
    return weight_kg / height_m ** 2


def bmi_category(bmi: float) -> str:
    """WHO adult categories. Each check starts where the previous one ends, so there are no gaps."""
    if bmi < 18.5:
        return "Underweight"
    if bmi < 25:
        return "Normal weight"
    if bmi < 30:
        return "Overweight"
    return "Obese"


def main() -> None:
    weight = read_number("Enter your weight in kg: ", 2, 650)
    height_cm = read_number("Enter your height in cm: ", 50, 275)

    bmi = calculate_bmi(weight, height_cm / 100)
    print(f"\nYour BMI is {bmi:.1f}: {bmi_category(bmi)}.")


if __name__ == "__main__":
    main()
```
