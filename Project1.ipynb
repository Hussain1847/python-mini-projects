# 📝 Project 1: Mad Libs

A word game: you type in a few words and the program drops them into a short story.

**Concepts:** `input()`, f-strings, small reusable functions, input validation.

### Three ways to join strings
Python has several ways to build a string from variables. f-strings (the last one) are the most readable, so they're used in the game below.

```python
channel = "PIAIC"

print("Subscribe to " + channel)            # 1. concatenation with +
print("Subscribe to {}".format(channel))    # 2. str.format()
print(f"Subscribe to {channel}")            # 3. f-string (recommended)
```

### The game

```python
def ask(prompt: str) -> str:
    """Keep asking until the user types something that isn't blank."""
    while True:
        answer = input(prompt).strip()
        if answer:
            return answer
        print("Please type at least one character.")


def build_story(adjective: str, verb1: str, verb2: str, famous_person: str) -> str:
    """Drop the user's words into the story template."""
    return (
        f"Computer programming is so {adjective}! "
        f"It makes me so excited all the time because I love to {verb1}. "
        f"Stay hydrated and {verb2} like you are {famous_person}!"
    )


def main() -> None:
    print("🎉 Welcome to Mad Libs!\n")
    adjective = ask("Adjective: ")
    verb1 = ask("Verb: ")
    verb2 = ask("Another verb: ")
    famous_person = ask("Famous person: ")
    print("\n" + build_story(adjective, verb1, verb2, famous_person))


if __name__ == "__main__":
    main()
```
